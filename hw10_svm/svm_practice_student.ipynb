## Коротко о теории

### Разделяющая гиперплоскость и margin

SVM ищет гиперплоскость $\mathbf{w}^\top \mathbf{x} + b = 0$, максимизирующую зазор (margin):

$$\text{margin} = \frac{2}{\|\mathbf{w}\|}$$

Это эквивалентно задаче:

$$\min_{\mathbf{w},b}\; \frac{1}{2}\|\mathbf{w}\|^2 + C\sum_i \xi_i \quad \text{при} \quad y_i(\mathbf{w}^\top\mathbf{x}_i + b) \geq 1 - \xi_i$$

### Ядровой трюк

RBF ядро: $K(\mathbf{x}_i, \mathbf{x}_j) = \exp(-\gamma\|\mathbf{x}_i - \mathbf{x}_j\|^2)$ — позволяет строить нелинейные границы без явного перехода в высокомерное пространство.

### Ключевые параметры

| Параметр | Влияние |
|---|---|
| `C` большой | Узкий margin, мало ошибок на train, риск переобучения |
| `C` малый | Широкий margin, допускает ошибки, лучше обобщение |
| `gamma` большой | Сложная граница, переобучение |
| `gamma` малый | Гладкая граница, недообучение |

## Шаг 6. Граница решений на двух признаках

Для визуализации возьмём два признака: `'mean radius'` и `'mean texture'`.

Обучите SVM с RBF ядром и лучшими найденными `C` и `gamma` на этих двух признаках, затем визуализируйте границу решений.

Ниже дана вспомогательная функция — вам нужно её вызвать.

````python
def plot_decision_boundary(model, X_2d, y, title='Граница решений SVM'):
    """Визуализирует границу решений обученной модели на двумерных данных."""
    h = 0.05
    x_min, x_max = X_2d[:, 0].min() - 0.5, X_2d[:, 0].max() + 0.5
    y_min, y_max = X_2d[:, 1].min() - 0.5, X_2d[:, 1].max() + 0.5
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                          np.arange(y_min, y_max, h))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(8, 6))
    plt.contourf(xx, yy, Z, alpha=0.25,
                 cmap=mcolors.ListedColormap(['tomato', 'steelblue']))
    plt.contour(xx, yy, Z, colors='k', linewidths=0.8)
    colors = ['tomato' if c == 0 else 'steelblue' for c in y]
    plt.scatter(X_2d[:, 0], X_2d[:, 1], c=colors, edgecolors='k', s=40, alpha=0.8)
    plt.xlabel('mean radius (scaled)')
    plt.ylabel('mean texture (scaled)')
    plt.title(title)
    plt.tight_layout()
    plt.show()
````

````python
# Выбираем два признака
feat1, feat2 = 'mean radius', 'mean texture'

X_train_2d = X_train[[feat1, feat2]].values
X_test_2d  = X_test[[feat1, feat2]].values

# YOUR CODE HERE
# 1) Создайте Pipeline: StandardScaler + SVC(kernel='rbf', C=<лучший C>, gamma=<лучший gamma>)
# 2) Обучите на X_train_2d, y_train
# 3) Вызовите plot_decision_boundary на тренировочных данных
# 4) Вызовите plot_decision_boundary на тестовых данных

pipe_2d = Pipeline([
    ('scaler', StandardScaler()),   
    ('svc', SVC(kernel='rbf', C=best_C, gamma=best_gamma))
])

pipe_2d.fit(X_train_2d, y_train.values)


plot_decision_boundary(pipe_2d, X_train_2d, y_train.values,
                       title='Train: Граница решений (2 признака)')
plot_decision_boundary(pipe_2d, X_test_2d, y_test.values,
                       title='Test: Граница решений (2 признака)')
````

### Задание B: влияние `C`

Проверим, как меняются train F1 и test F1 при фиксированных `kernel='rbf'` и `gamma='scale'`. Чем больше `C`, тем сильнее модель штрафует ошибки на train и тем уже становится margin.

````python
C_values = np.logspace(-3, 3, 13)
C_scores = []

for C in C_values:
    model = Pipeline([
        ('scaler', StandardScaler()),
        ('svm', SVC(kernel='rbf', C=C, gamma='scale', random_state=42))
    ])
    model.fit(X_train, y_train)

    train_f1 = f1_score(y_train, model.predict(X_train))
    test_f1 = f1_score(y_test, model.predict(X_test))
    C_scores.append({'C': C, 'train_f1': train_f1, 'test_f1': test_f1})

C_scores = pd.DataFrame(C_scores)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.semilogx(C_scores['C'], C_scores['train_f1'], marker='o', label='train F1')
ax.semilogx(C_scores['C'], C_scores['test_f1'], marker='o', label='test F1')
ax.set_xlabel('C')
ax.set_ylabel('F1')
ax.set_title("Влияние C при kernel='rbf', gamma='scale'")
ax.legend()
plt.tight_layout()
plt.show()

display(C_scores.round(4))

best_row = C_scores.loc[C_scores['test_f1'].idxmax()]
C_scores['gap'] = C_scores['train_f1'] - C_scores['test_f1']
print(f"Лучший test F1 при C={best_row['C']:.4g}: {best_row['test_f1']:.4f}")
print('При больших C train F1 растёт до 1.0, а test F1 падает — это признак переобучения.')
````

**Ответ по B:** лучшее качество на test получается около `C=1`. После `C≈10-30` train F1 продолжает расти почти до `1.0`, а test F1 снижается, поэтому там уже начинается переобучение.

````python
balanced_model = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(
        kernel=best_kernel,
        C=best_C,
        gamma=best_gamma,
        class_weight='balanced',
        random_state=42
    ))
])

balanced_model.fit(X_train, y_train)
y_pred_balanced = balanced_model.predict(X_test)

balanced_compare = pd.DataFrame([
    {
        'model': 'best_grid',
        'accuracy': accuracy_score(y_test, y_pred_best),
        'f1': f1_score(y_test, y_pred_best),
        'recall_malignant': recall_score(y_test, y_pred_best, pos_label=0),
    },
    {
        'model': 'balanced',
        'accuracy': accuracy_score(y_test, y_pred_balanced),
        'f1': f1_score(y_test, y_pred_balanced),
        'recall_malignant': recall_score(y_test, y_pred_balanced, pos_label=0),
    },
]).set_index('model')

display(balanced_compare.round(4))
print(classification_report(y_test, y_pred_balanced, target_names=data.target_names))
````

**Ответ по C:** в этом разбиении `class_weight='balanced'` не улучшает recall для `malignant`: он остаётся примерно тем же, а общие метрики немного снижаются. Но в медицинских задачах этот recall всё равно важно контролировать отдельно, потому что ошибка “злокачественная опухоль предсказана как доброкачественная” может привести к пропущенному лечению.

### Задание D*: `RandomizedSearchCV`

Проверим случайный поиск по более широкой области `C` и `gamma`, затем сравним его с обычным `GridSearchCV` по времени, CV F1 и test F1.

````python
import time
from scipy.stats import loguniform
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    'svm__C': loguniform(1e-3, 1e3),
    'svm__gamma': loguniform(1e-4, 1e1),
    'svm__kernel': ['linear', 'rbf'],
}

random_search = RandomizedSearchCV(
    pipe_scaler_svm,
    param_distributions=param_dist,
    n_iter=30,
    cv=5,
    scoring='f1',
    random_state=42
)

start = time.time()
random_search.fit(X_train, y_train)
random_time = time.time() - start

y_pred_random = random_search.best_estimator_.predict(X_test)

random_compare = pd.DataFrame([
    {
        'search': 'GridSearchCV',
        'best_cv_f1': search.best_score_,
        'test_f1': f1_score(y_test, y_pred_best),
        'time_sec': np.nan,
        'best_params': search.best_params_,
    },
    {
        'search': 'RandomizedSearchCV',
        'best_cv_f1': random_search.best_score_,
        'test_f1': f1_score(y_test, y_pred_random),
        'time_sec': random_time,
        'best_params': random_search.best_params_,
    },
])

display(random_compare)
print('Лучшие параметры RandomizedSearchCV:', random_search.best_params_)
````

**Ответ по D:** `RandomizedSearchCV` проверяет не всю сетку, а случайные комбинации из более широкой области. На этом датасете качество получается примерно таким же, как у `GridSearchCV`, но результат зависит от `random_state` и `n_iter`. Такой подход полезен, когда параметров много и полный перебор становится слишком дорогим.
